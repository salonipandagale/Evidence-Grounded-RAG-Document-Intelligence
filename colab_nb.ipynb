```python
from pypdf import PdfReader
import os

def read_txt(path):
    with open(path, "r", encoding="utf-8", errors="ignore") as f:
        return f.read()


def read_pdf(path):
    reader = PdfReader(path)

    pages = []

    for page in reader.pages:
        pages.append(page.extract_text() or "")

    return "\n".join(pages)


def load_documents_from_folder(folder):
    documents = []

    for root, _, files in os.walk(folder):

        for filename in files:

            path = os.path.join(root, filename)

            if filename.lower().endswith(".txt"):
                text = read_txt(path)

            elif filename.lower().endswith(".pdf"):
                text = read_pdf(path)

            else:
                continue

            documents.append({
                "id": len(documents),
                "source": path,
                "text": text
            })

    return documents
```

## splitting documents into chunks

```python
import re

def simple_chunk(text, chunk_size=300, overlap=50):
    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    if not text:
        return []

    chunks = []
    start = 0

    while start < len(text):
        end = start + chunk_size

        chunk = text[start:end]
        chunks.append(chunk)

        start += max(1, chunk_size - overlap)

    return chunks
```

```python
#test if chunk is working
text = documents[0]["text"]

chunks = simple_chunk(
    text,
    chunk_size=300,
    overlap=50
)

print("Number of chunks:", len(chunks))

for i, chunk in enumerate(chunks):
    print(f"\n--- Chunk {i} ---")
    print(chunk)
```

```python
from dataclasses import dataclass

@dataclass
class Chunk:
    chunk_id: int
    doc_id: int
    source: str
    text: str
```

```python
def make_chunks(docs, chunk_size=300, overlap=50):
    chunks = []

    for doc in docs:

        text_chunks = simple_chunk(
            doc["text"],
            chunk_size=chunk_size,
            overlap=overlap
        )

        for text in text_chunks:

            chunks.append(
                Chunk(
                    chunk_id=len(chunks),
                    doc_id=doc["id"],
                    source=doc["source"],
                    text=text
                )
            )

    return chunks
```

```python
chunks = make_chunks(
    documents,
    chunk_size=300,
    overlap=50
)

print("Total chunks:", len(chunks))

for chunk in chunks:
    print("\nChunk ID:", chunk.chunk_id)
    print("Document ID:", chunk.doc_id)
    print("Source:", chunk.source)
    print("Text:", chunk.text)
```

## encode all chunks

```python
import numpy as np

chunk_texts = [chunk.text for chunk in chunks]

embeddings = embedder.encode(
    chunk_texts,
    convert_to_numpy=True,
    show_progress_bar=True
)

print("Embedding shape:", embeddings.shape)
print("Data type:", embeddings.dtype)
```

```python
import faiss

embeddings = embeddings.astype("float32")

faiss.normalize_L2(embeddings)

print("Normalized embedding shape:", embeddings.shape)
print("First embedding norm:", np.linalg.norm(embeddings[0]))
```

```python
embedding_dim = embeddings.shape[1]

index = faiss.IndexFlatIP(embedding_dim)

index.add(embeddings)

print("FAISS index created!")
print("Embedding dimension:", embedding_dim)
print("Number of vectors in index:", index.ntotal)
```

```python
def search_chunks(query, top_k=3):
    query_embedding = embedder.encode(
        [query],
        convert_to_numpy=True
    ).astype("float32")

    faiss.normalize_L2(query_embedding)

    scores, indices = index.search(query_embedding, top_k)

    results = []

    for score, idx in zip(scores[0], indices[0]):
        if idx == -1:
            continue

        results.append({
            "score": float(score),
            "chunk": chunks[idx]
        })

    return results
```

```python
query = "How many annual leave days do full-time employees get?"

results = search_chunks(query, top_k=3)

for result in results:
    print("\nSimilarity score:", result["score"])
    print("Chunk ID:", result["chunk"].chunk_id)
    print("Text:", result["chunk"].text)
```

```python
class RagIndex:
    def __init__(self, chunks, embedder):
        self.chunks = chunks
        self.embedder = embedder

        # Create embeddings for all chunks
        texts = [chunk.text for chunk in chunks]

        embeddings = self.embedder.encode(
            texts,
            convert_to_numpy=True,
            show_progress_bar=True
        ).astype("float32")

        # Normalize for cosine similarity
        faiss.normalize_L2(embeddings)

        # Build FAISS index
        self.index = faiss.IndexFlatIP(embeddings.shape[1])
        self.index.add(embeddings)

    def search(self, query, top_k=3):
        # Convert query into embedding
        query_embedding = self.embedder.encode(
            [query],
            convert_to_numpy=True
        ).astype("float32")

        # Normalize query
        faiss.normalize_L2(query_embedding)

        # Search FAISS
        scores, indices = self.index.search(
            query_embedding,
            top_k
        )

        results = []

        for score, idx in zip(scores[0], indices[0]):
            if idx == -1:
                continue

            results.append({
                "score": float(score),
                "chunk": self.chunks[idx]
            })

        return results
```

```python
rag_index = RagIndex(
    chunks=chunks,
    embedder=embedder
)

print("RAG index ready!")
print("Number of chunks:", len(rag_index.chunks))
print("Vectors in FAISS:", rag_index.index.ntotal)
```

```python
def format_context(results):
    context_parts = []

    for result in results:
        chunk = result["chunk"]

        context_parts.append(
            f"[Source: {chunk.source} | Chunk: {chunk.chunk_id}]\n"
            f"{chunk.text}"
        )

    return "\n\n".join(context_parts)
```

```python
query = "How many sick leave days are available to full-time employees?"

results = rag_index.search(query, top_k=2)

context = format_context(results)

print(context)
```

```python
def build_rag_prompt(question, context):
    prompt = f"""
Answer the question using ONLY the context provided below.

If the context does not contain the answer, say:
"I don't know from the provided documents."

Context:
{context}

Question:
{question}

Answer:
"""

    return prompt.strip()
```

```python
prompt = build_rag_prompt(
    question=query,
    context=context
)

print(prompt)
```

```python
def rag_answer(question, top_k=3, max_new_tokens=100):
    # 1. Retrieve relevant chunks
    results = rag_index.search(
        question,
        top_k=top_k
    )

    # 2. Format retrieved chunks
    context = format_context(results)

    # 3. Build grounded prompt
    prompt = build_rag_prompt(
        question,
        context
    )

    # 4. Tokenize prompt
    inputs = gen_tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True
    )

    # 5. Generate answer
    outputs = gen_model.generate(
        **inputs,
        max_new_tokens=max_new_tokens
    )

    # 6. Decode answer
    answer = gen_tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return {
        "question": question,
        "answer": answer,
        "results": results,
        "context": context
    }
```

## Improve Retrieval With a Similarity Threshold

```python
class RagIndex:
    def __init__(self, chunks, embedder):
        self.chunks = chunks
        self.embedder = embedder

        texts = [chunk.text for chunk in chunks]

        embeddings = self.embedder.encode(
            texts,
            convert_to_numpy=True,
            show_progress_bar=True
        ).astype("float32")

        faiss.normalize_L2(embeddings)

        self.index = faiss.IndexFlatIP(embeddings.shape[1])
        self.index.add(embeddings)

    def search(self, query, top_k=3, score_threshold=0.60):
        query_embedding = self.embedder.encode(
            [query],
            convert_to_numpy=True
        ).astype("float32")

        faiss.normalize_L2(query_embedding)

        scores, indices = self.index.search(
            query_embedding,
            top_k
        )

        results = []

        for score, idx in zip(scores[0], indices[0]):
            if idx == -1:
                continue

            if score < score_threshold:
                continue

            results.append({
                "score": float(score),
                "chunk": self.chunks[idx]
            })

        return results
```

```python
rag_index = RagIndex(
    chunks=chunks,
    embedder=embedder
)

print("RAG index rebuilt!")
print("Vectors:", rag_index.index.ntotal)
```

```python
def show_evidence(results):
    if not results:
        print("No relevant evidence found.")
        return

    print("Retrieved Evidence")
    print("=" * 60)

    for i, result in enumerate(results, start=1):
        chunk = result["chunk"]

        print(f"\nEvidence {i}")
        print("-" * 60)
        print("Source:", os.path.basename(chunk.source))
        print("Document ID:", chunk.doc_id)
        print("Chunk ID:", chunk.chunk_id)
        print("Similarity:", round(result["score"], 4))
        print("\nText:")
        print(chunk.text)
```

```python
#combined response

def rag_answer(
    question,
    top_k=3,
    score_threshold=0.60,
    max_new_tokens=100
):
    # 1. Retrieve evidence
    results = rag_index.search(
        question,
        top_k=top_k,
        score_threshold=score_threshold
    )

    # 2. No relevant evidence
    if not results:
        return {
            "question": question,
            "answer": "I don't know from the provided documents.",
            "results": [],
            "context": ""
        }

    # 3. Build context
    context = format_context(results)

    # 4. Build grounded prompt
    prompt = build_rag_prompt(
        question,
        context
    )

    # 5. Tokenize
    inputs = gen_tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True
    )

    # 6. Generate answer
    outputs = gen_model.generate(
        **inputs,
        max_new_tokens=max_new_tokens
    )

    # 7. Decode
    answer = gen_tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return {
        "question": question,
        "answer": answer,
        "results": results,
        "context": context
    }
```

## Improve the Chunking Strategy

```python
import re

def sentence_chunk(text, chunk_size=300, overlap_sentences=1):
    text = re.sub(r"\s+", " ", text).strip()

    if not text:
        return []

    # Split into sentences
    sentences = re.split(r"(?<=[.!?])\s+", text)

    chunks = []
    current_sentences = []
    current_length = 0

    for sentence in sentences:
        sentence = sentence.strip()

        if not sentence:
            continue


        if (
            current_sentences
            and current_length + len(sentence) > chunk_size
        ):
            chunks.append(" ".join(current_sentences))

            # Keep the last few sentences as overlap
            current_sentences = current_sentences[
                -overlap_sentences:
            ]

            current_length = sum(
                len(s) for s in current_sentences
            )

        current_sentences.append(sentence)
        current_length += len(sentence) + 1

    # Add final chunk
    if current_sentences:
        chunks.append(" ".join(current_sentences))

    return chunks
```
